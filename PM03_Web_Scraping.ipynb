```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
url = "https://books.toscrape.com/"
response = requests.get(
    url,
    timeout=20
)
print(response.status_code)
```

```python
response.encoding = "utf-8-sig"

Soup = BeautifulSoup(
    response.text,
    "html.parser"
)
print(Soup.title.text)
```

```python
Books = Soup.select(
    "article.product_pod"
)
print("Number of books found:", len(Books))
```

```python
for book in Books:
  title = book.h3.a['title']
  print(title)
```

```python
for book in Books:
  price = book.select_one(
      ".price_color"
  ).get_text(strip=True)
  print(price)
```

```python
for book in Books:
  rating = book.select_one(
      ".star-rating"
  )["class"][1]
  print(rating)
```

```python
for book in Books:
  availability = book.select_one(
      ".availability"
  ).get_text(strip=True)
  print(availability)
```

```python
data = []

url = "https://books.toscrape.com/"

while url:

  response = requests.get(url, timeout=20)
  response.encoding = "utf-8-sig"

  Soup = BeautifulSoup(response.text, "html.parser")

  Books = Soup.select("article.product_pod")

  for book in Books:

    title = book.h3.a["title"]

    price = book.select_one(
        ".price_color"
    ).get_text(strip=True)

    price = float(price.replace("£", ""))

    rating = book.select_one(
        ".star-rating"
    )["class"][1]

    availability = book.select_one(
        ".availability"
    ).get_text(strip=True)

    data.append ({
        "Book Title": title,
        "Price": price,
        "Rating": rating,
        "Availability": availability
    })

  next_button = Soup.select_one("li.next a")

  if next_button:
      url = requests.compat.urljoin(url, next_button["href"])
  else:
      url = None

df = pd.DataFrame(data)

print("Total books scraped:", len(df))

display(df)
```

```python
df.to_csv(
    "Scraped_books.csv",
    index = False,
    encoding = "utf-8-sig"
)
print("CSV file Saved successfully!")
```

```python
df.to_excel(
    "Scraped_books.xlsx",
    index = False
)
print("Excel file saved successfully!")
```
